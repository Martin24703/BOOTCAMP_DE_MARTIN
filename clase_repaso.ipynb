La lista de animales

Un zoológico quiere mostrar todos sus animales.

🎯 Tu misión

Crear una lista con 5 animales

Usar for para imprimir cada animal


```python
zoologico = ['panda', 'ornitorrinco', 'elefante', 'puma', 'tigre']
for i in range(5):
    print(zoologico[i])
```
