Crear dos métodos para nuestro animal y ejecutar, debe ser algo específico de ese animal.

Ejemplo: "Cuando hablo, digo X".
Ejemplo: "Cuando me muevo, hago X"

```python
class animal:
    def __init__(self, nombre, color):
        self.nombre = nombre
        self.color = color
    
    def ruido1(self, ruido):
        print(f'soy un {self.nombre} de color {self.color} y cuando quiero hacer un ruido hago un {ruido}')

    def comer1(self, comer):
        print(f'soy un {self.nombre} de color {self.color} y cuando quiero comer agarro una planta con {comer} la arranco y la llevo a mi boca')
    
mi_elefante = animal('elefante', 'gris')

mi_elefante.ruido1('soplido con mi trompa')

mi_elefante.comer1('la trompa') 
```
