## Challenge 1 🤺
Escribir un programa que cree un `diccionario vacío` y lo vaya llenado con información sobre una persona (por ejemplo nombre, edad, sexo, teléfono, correo electrónico, etc.) que se le `pida al usuario`. 

Cada vez que se `añada` un nuevo dato debe `imprimirse` el contenido del diccionario.


```python
biografia = {}

def biografia_del_usuario():
    biografia['nombre'] = input('agregue su nombre')
    print(biografia)
    biografia['apellido'] = input('agregue su apellido')
    print(biografia)
    biografia['edad'] = int(input('ingrese su edad'))
    print(biografia)
    biografia['sexo'] = input('declare su sexo')
    print(biografia)
    biografia['correo electronico'] = input('ingrese su correo electronico')
    print(biografia)
    biografia['numero de telefono'] = int(input('ingrese su numero de telefono'))
    print(biografia)

biografia_del_usuario()
```
